#### Q35
##### import copy

##### a = [[1, 2], [3, 4]]
##### b = copy.copy(a)
##### b[0][1] = 99
##### print(a)

```python
## Expected Outcome : [[1, 99], [3, 4]]
## this above question refers to shallow copy concept
## a shallow copy creates a new outer list but it shares the inner list
## means if we modify an inner list it changes the original list too
```

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
```

#### Q36
##### import copy

##### a = [[1, 2], [3, 4]]
##### b = copy.deepcopy(a)
##### b[0][1] = 99
##### print(a)

```python
## Expected Outcome : [[1, 2], [3, 4]]
## the above wuestion refers to the deep copy concept
## deepcopy duplicates the nested lists too, rather than sharing them
## so changing b doesn't affect a 
```

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
```
