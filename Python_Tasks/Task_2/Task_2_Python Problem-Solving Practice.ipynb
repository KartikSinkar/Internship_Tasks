#### Write a Python program to find the longest word that appears exactly once in a sentence. If multiple qualifying words have the same maximum length, return the first one encountered. Treat uppercase and lowercase versions as the same word.
#### Example input:
#### Python makes coding useful Python
#### Expected output:
#### coding

1. OBSERVE
- Input: a sentence. Expected output: the longest word that occurs exactly once.

2. UNDERSTAND
- Treat uppercase and lowercase as the same.
- If qualifying words tie in length, return the first one.

3. BREAK
- Convert to lowercase.
- Split into words.
- Keep words occurring once.
- Find the longest among them.

4. LOGIC
- Build a list of unique-occurrence words, then update the longest result only when a strictly longer word is found.

5. CODE 

```python
sentence = input("Enter a sentence: ").lower()
words = sentence.split()
unique_words = []

for word in words:
    if words.count(word) == 1:
        unique_words.append(word)
if unique_words:
    longest = unique_words[0]
    for word in unique_words:
        if len(word) > len(longest):
            longest = word
    print(longest)
else:
    print("No unique word found")
```
